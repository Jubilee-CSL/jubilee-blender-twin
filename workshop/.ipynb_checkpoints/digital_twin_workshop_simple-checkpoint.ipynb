## 0 · Setup

```python
import re
import shutil
import subprocess
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

TWIN = Path.cwd().parent                                   # jubilee-blender-twin/
ADDON = TWIN / "blender_addon" / "jubilee_digital_twin"    # the twin's own Blender scripts
PIPE = TWIN / "pipeline_data"                              # where the twin keeps its scenes and CSVs
OUT = Path.cwd() / "out_simple"
OUT.mkdir(exist_ok=True)
Path("scripts").mkdir(exist_ok=True)


def blender(blend, *args):
    # Run Blender without a window on `blend`, return what it printed.
    cmd = ["blender", str(blend), "--background", *map(str, args)]
    run = subprocess.run(cmd, capture_output=True, text=True, encoding="utf-8", errors="replace")
    if run.returncode != 0:
        print(run.stdout[-3000:], run.stderr[-3000:])
        raise RuntimeError("Blender failed")
    return run.stdout


def show(*paths, titles=None, width=5):
    fig, axes = plt.subplots(1, len(paths), figsize=(width * len(paths), width * 0.6), squeeze=False)
    for i, (ax, p) in enumerate(zip(axes[0], paths)):
        ax.imshow(Image.open(p)); ax.axis("off")
        if titles: ax.set_title(titles[i])
    plt.tight_layout()
```

The working scene has the machine and the calibrated `Toolhead_Cam`, but nothing on the bed. This Blender script takes the plant from `basil_3D.blend`, parents it to the bed plate (so it moves with Z), and gives the camera a real lens with a fixed focus distance.

```python
%%writefile scripts/build_scene.py
# Runs inside Blender: put a plant on the bed and give the camera a real lens.
import sys
import bpy

plant_blend, out = sys.argv[sys.argv.index("--") + 1:]
plate = bpy.data.objects["z_plate_3_V5"]            # the bed plate, moved by Z

existing = set(bpy.data.objects.keys())
with bpy.data.libraries.load(plant_blend) as (src, dst):
    dst.objects = [name for name in src.objects if name not in existing]

def on_bed(obj):
    parent = obj.parent
    while parent:
        if parent.name.startswith("z_plate_3_V5"):
            return True
        parent = parent.parent
    return False

plant = [o for o in dst.objects if o and on_bed(o) and not o.name.startswith("z_plate")]
for obj in plant:
    if obj.parent.name.startswith("z_plate"):       # re-attach to this scene's bed, same place
        inverse = obj.matrix_parent_inverse.copy()
        obj.parent = plate
        obj.matrix_parent_inverse = inverse
    bpy.context.scene.collection.objects.link(obj)
bpy.context.scene["plant"] = [o.name for o in plant]

lens = bpy.data.objects["Toolhead_Cam"].data        # fixed focus: moving Z changes sharpness
lens.dof.use_dof = True
lens.dof.aperture_fstop = 4.0
lens.dof.focus_distance = 0.14                      # metres

bpy.context.scene.render.resolution_percentage = 50
bpy.data.orphans_purge(do_recursive=True)
bpy.ops.wm.save_as_mainfile(filepath=out)
print("plant objects:", [o.name for o in plant])
```

```python
SCENE = PIPE / "workshop_simple.blend"
log = blender(PIPE / "jubilee_working.blend", "--python", "scripts/build_scene.py", "--",
              TWIN / "blender_models" / "basil_3D.blend", SCENE)
print([line for line in log.splitlines() if "plant objects" in line][0])
print("Scene saved:", SCENE)
```

First picture, with the twin's own `snapshot.py`. **X/Y** move the camera, **Z** moves the bed: bigger Z means the bed is lower.

```python
blender(SCENE, "--python", ADDON / "snapshot.py", "--", "--x", 147, "--y", 145, "--z", 310,
        "--output", OUT / "first.jpg")
show(OUT / "first.jpg", width=8)
```

## 1 · Check the plan before running it

This raster was written for a flat Petri dish, with the bed raised to Z = 150. Today the basil is on the bed. **What will happen?**

```python
PLAN = '''G90
G0 X100 Y100 Z150
G1 X200 Y100
G1 X200 Y145
G1 X100 Y145
G1 X100 Y190
G1 X200 Y190'''
```

**Step 1: G-code → path.** The twin's parser turns the plan into one machine position every 10 mm.
It always starts at home (0, 0, 0), which with this plant is *inside* the leaves. On the real machine you load the plant with the bed already lowered, so we start at the first commanded position.

```python
import contextlib, io
from jubilee_twin.pipeline.path_follower import build_path

def plan_to_path(plan, step_mm=10):
    lines = plan.splitlines()
    with contextlib.redirect_stdout(io.StringIO()):          # the parser prints debug lines
        path = np.array(build_path(lines, step_mm))
        to_start = build_path(lines[:2], step_mm)            # home -> first position
    path = path[len(to_start) - 1:]
    np.savetxt(PIPE / "pathout.csv", path, delimiter=",", fmt="%.3f")   # the file the twin animates
    return path

path = plan_to_path(PLAN)
print(len(path), "frames, first ones:\n", path[:4, :3])
```

**Step 2: animate + ray-trace**, with the twin's `animate_path.py` and `ray_tracing.py`, on a copy of the scene. Blender saves the animated copy, so you can open it and watch.

```python
def check(plan, name):
    path = plan_to_path(plan)
    saved = PIPE / f"check_{name}.blend"
    shutil.copy(SCENE, saved)
    log = blender(saved, "--python", ADDON / "animate_path.py", "--python", ADDON / "ray_tracing.py")
    hits = [(int(frame), obj) for obj, frame in re.findall(r"Collision between gantry and (.+) during frame (\d+)", log)]
    print(f"{len(hits)} collision frame(s)")
    for frame, obj in hits[:5]:
        x, y, z = path[frame - 1, :3]
        print(f"  frame {frame:3d}: gantry hits {obj!r} at X={x:.0f} Y={y:.0f} Z={z:.0f}")
    print(f"\nAnimated scene saved: {saved}")
    print("Open it in Blender, press Space to play, or jump to the frames listed above.")
    return hits, saved

hits, saved = check(PLAN, "petri_plan")
```

## 2 · Autofocus with the virtual camera

The camera has a fixed focus distance, so moving the bed in Z moves the plant in and out of focus. We take a **Z-stack** with the twin's virtual scanner (`scan.py`, the same as `jubilee-twin scan`), then let a sharpness score pick the best Z.

```python
X, Y = 147, 145
blender(SCENE, "--python", ADDON / "scan.py", "--",
        "--x-min", X, "--x-max", X, "--x-steps", 1,
        "--y-min", Y, "--y-max", Y, "--y-steps", 1,
        "--z-min", 220, "--z-max", 340, "--z-steps", 9,
        "--width", 960, "--height", 528, "--output-root", OUT / "zstack")

scan_dir = sorted((OUT / "zstack").iterdir())[-1]           # newest scan folder
images = sorted(scan_dir.glob("*.jpg"), key=lambda p: int(p.stem.split("_z")[1]))
zs = [int(p.stem.split("_z")[1]) for p in images]
print("Saved in", scan_dir)
```

**Sharpness = variance of the Laplacian**: a sharp image has strong edges, so its second derivative varies a lot.

```python
from scipy import ndimage

def sharpness(path):
    gray = np.asarray(Image.open(path).convert("L"), float)
    return ndimage.laplace(gray).var()

scores = [sharpness(p) for p in images]
best = int(np.argmax(scores))
print("Best focus: Z =", zs[best])

plt.figure(figsize=(5, 3)); plt.plot(zs, scores, "-o"); plt.axvline(zs[best], color="crimson", ls="--")
plt.xlabel("bed Z (mm)"); plt.ylabel("sharpness")
show(images[0], images[best], titles=[f"Z = {zs[0]}", f"Z = {zs[best]} (chosen)"])
```

## 3 · Change lighting and background

The scene is lit by an **HDRI**, a 360° photo that is *both* the light and the background (the wooden floor is the kitchen HDRI). Put `.hdr` files in `workshop/hdri/` (e.g. from polyhaven.com).

**In Blender (demo):** open the scene → **World** tab (red globe) → *Color* → **Environment Texture** → *Open* an `.hdr`. *Strength* sets the intensity. To rotate it, add *Texture Coordinate* → *Mapping* in the World node editor (Shading workspace) and change *Rotation Z*. **F12** renders through the toolhead camera.

**From Python**, the same three settings:

```python
%%writefile scripts/lighting.py
# Runs inside Blender: render the same view under several HDRIs.
import sys
from pathlib import Path
import bpy

sys.path.insert(0, str(Path(__file__).resolve().parents[2] / "blender_addon" / "jubilee_digital_twin"))
import scene_utils

x, y, z, strength, rotation, out_dir, *hdris = sys.argv[sys.argv.index("--") + 1:]
scene_utils.drive_to_mm(float(x), float(y), float(z))
bpy.context.scene.render.image_settings.file_format = "PNG"

nodes = bpy.context.scene.world.node_tree.nodes
links = bpy.context.scene.world.node_tree.links
env = next(n for n in nodes if n.type == "TEX_ENVIRONMENT")
background = next(n for n in nodes if n.type == "BACKGROUND")

background.inputs["Strength"].default_value = float(strength)       # how bright
mapping = nodes.new("ShaderNodeMapping")                            # where the light comes from
coords = nodes.new("ShaderNodeTexCoord")
links.new(coords.outputs["Generated"], mapping.inputs["Vector"])
links.new(mapping.outputs["Vector"], env.inputs["Vector"])
mapping.inputs["Rotation"].default_value[2] = float(rotation) * 3.14159 / 180

for hdri in hdris:                                                  # which place it is
    env.image = bpy.data.images.load(hdri, check_existing=True)
    bpy.context.scene.render.filepath = str(Path(out_dir) / (Path(hdri).stem + ".png"))
    bpy.ops.render.render(write_still=True)
```

```python
HDRIS = sorted(Path("hdri").glob("*.hdr"))
STRENGTH, ROTATION = 3.0, 0      # ✏️ try 0.8 and 120

blender(SCENE, "--python", "scripts/lighting.py", "--", X, Y, zs[best], STRENGTH, ROTATION,
        OUT / "lighting", *[h.resolve() for h in HDRIS])
show(OUT / "first.jpg", *[OUT / "lighting" / f"{h.stem}.png" for h in HDRIS],
     titles=["kitchen (default)", *[h.stem for h in HDRIS]], width=4.5)
```

The plant, the camera and the geometry did not move: only the appearance changed. Every image has **the same depth map, the same mask, the same camera pose**. Labels come for free, which is what Exercise 4 uses.

## 4 · MoGe-2: reconstruct from one image, and score it

[MoGe-2](https://huggingface.co/spaces/Ruicheng/MoGe-2) predicts a 3D point for every pixel of a single image. On a real plant nobody knows the true shape. In the twin we do: shoot one ray per pixel through the camera and record what it hits.

```python
%%writefile scripts/ground_truth.py
# Runs inside Blender: render one image + its exact 3D ground truth (ray cast, half resolution).
import sys
from pathlib import Path
import bpy
import numpy as np
from mathutils import Vector

sys.path.insert(0, str(Path(__file__).resolve().parents[2] / "blender_addon" / "jubilee_digital_twin"))
import scene_utils

x, y, z, out_dir = sys.argv[sys.argv.index("--") + 1:]
out_dir = Path(out_dir); out_dir.mkdir(parents=True, exist_ok=True)
scene = bpy.context.scene
scene_utils.drive_to_mm(float(x), float(y), float(z))
scene.render.image_settings.file_format = "PNG"
scene.render.filepath = str(out_dir / "moge_input.png")
bpy.ops.render.render(write_still=True)

W = scene.render.resolution_x * scene.render.resolution_percentage // 200
H = scene.render.resolution_y * scene.render.resolution_percentage // 200
depsgraph = bpy.context.evaluated_depsgraph_get()
cam = scene.camera.evaluated_get(depsgraph)
to_cam = cam.matrix_world.inverted()

# The camera's image plane at distance 1, from Blender itself (handles lens shift).
tr, br, bl, tl = cam.data.view_frame(scene=scene)
dist = -tr.z
fx, fy = W * dist / (tr.x - bl.x), H * dist / (tr.y - br.y)
cx, cy = -bl.x * W / (tr.x - bl.x) - 0.5, tr.y * H / (tr.y - br.y) - 0.5
K = np.array([[fx, 0, cx], [0, fy, cy], [0, 0, 1]])

plant = set(scene["plant"])
points = np.zeros((H, W, 3), np.float32)          # OpenCV camera frame: x right, y down, z forward
mask = np.zeros((H, W), bool)
for v in range(H):
    for u in range(W):
        ray = cam.matrix_world.to_3x3() @ Vector(((u - cx) / fx * dist, -(v - cy) / fy * dist, -dist))
        hit, location, _, _, obj, _ = scene.ray_cast(depsgraph, cam.matrix_world.translation, ray.normalized())
        if hit:
            p = to_cam @ location
            points[v, u] = (p.x, -p.y, -p.z)
            mask[v, u] = obj.original.name in plant
np.savez(out_dir / "ground_truth.npz", points=points, mask=mask, K=K)
```

```python
EX4 = OUT / "moge"
blender(SCENE, "--python", "scripts/ground_truth.py", "--", X, Y, zs[best], EX4)   # ~30 s
gt = np.load(EX4 / "ground_truth.npz")
points, mask, K = gt["points"], gt["mask"], gt["K"]

fig, ax = plt.subplots(1, 3, figsize=(15, 3))
ax[0].imshow(Image.open(EX4 / "moge_input.png")); ax[0].set_title("upload this image")
ax[1].imshow(np.where(points[..., 2] > 0, points[..., 2], np.nan), cmap="turbo"); ax[1].set_title("true depth")
ax[2].imshow(mask, cmap="gray"); ax[2].set_title("plant mask")
for a in ax: a.axis("off")
print("Image to upload:", EX4 / "moge_input.png")
```

### Run MoGe-2 online
1. Open **https://huggingface.co/spaces/Ruicheng/MoGe-2**, upload `out_simple/moge/moge_input.png`, run.
2. Download **`pointcloud.ply`** right away (files are deleted after a few minutes) into `out_simple/moge/`.

```python
from plyfile import PlyData

ply = PlyData.read(EX4 / "pointcloud.ply")["vertex"]
moge = np.stack([ply["x"], ply["y"], ply["z"]], axis=1) * [1, -1, -1]   # demo flips y, z on export
print(len(moge), "MoGe points, same camera frame as the ground truth")
```

**Which MoGe points are on the plant?** Project each point into the image with the twin's `K` and look it up in the plant mask.

```python
def pixel_of(p):
    u = np.round(K[0, 0] * p[:, 0] / p[:, 2] + K[0, 2]).astype(int)
    v = np.round(K[1, 1] * p[:, 1] / p[:, 2] + K[1, 2]).astype(int)
    inside = (p[:, 2] > 0) & (u >= 0) & (u < mask.shape[1]) & (v >= 0) & (v < mask.shape[0])
    return u.clip(0, mask.shape[1] - 1), v.clip(0, mask.shape[0] - 1), inside

u, v, inside = pixel_of(moge)
on_plant = inside & mask[v, u]
true_plant = points[mask]

scale = np.median(true_plant[:, 2]) / np.median(moge[on_plant, 2])
print(f"MoGe-2 distance error: {(1 / scale - 1) * 100:+.1f} %   (MoGe-2 claims metric scale)")
```

**How good is the shape?** Remove that global scale, then measure, for every MoGe plant point, the distance to the nearest true point.

```python
from scipy.spatial import cKDTree

pred = moge[on_plant] * scale
error = cKDTree(true_plant).query(pred)[0] * 1000            # mm
print(f"median error {np.median(error):.1f} mm, {np.mean(error < 5) * 100:.0f} % of points within 5 mm")

fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].scatter(true_plant[::20, 2] * 100, true_plant[::20, 1] * 100, s=0.5, c="tab:green", label="twin (truth)")
ax[0].scatter(pred[::20, 2] * 100, pred[::20, 1] * 100, s=0.5, c="tab:orange", label="MoGe-2")
ax[0].invert_yaxis(); ax[0].set_xlabel("distance from camera (cm)"); ax[0].set_ylabel("y (cm)")
ax[0].set_title("side view"); ax[0].legend(markerscale=15)
ax[1].hist(error, bins=50, range=(0, 30)); ax[1].set_xlabel("error (mm)"); ax[1].set_title("error per MoGe point");
```
